```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# 1. Загрузка данных
# ============================================================
df = pd.read_excel('database-1-2026-09-24.xlsx', sheet_name='Лист1')

target = 'I_p'
features = ['D_n', 'p_nom', 'bar_e_0']

X = df[features].values
y = df[target].values

print("="*70)
print(f"ВЫВОД ФОРМУЛЫ ДЛЯ {target}")
print("="*70)

# ============================================================
# 2. МОДЕЛЬ 1: Линейная (без свободного члена и с ним)
# ============================================================
print("\n" + "─"*70)
print("МОДЕЛЬ 1: Линейная  I_p = a0 + a1·D_n + a2·p_nom + a3·bar_e_0")
print("─"*70)

m1 = LinearRegression().fit(X, y)
y1 = m1.predict(X)
print(f"R²  = {r2_score(y, y1):.6f}")
print(f"MAE = {mean_absolute_error(y, y1):.4f}")
print(f"RMSE= {np.sqrt(mean_squared_error(y, y1)):.4f}")
print(f"\nI_p = {m1.intercept_:.6f}")
for f, c in zip(features, m1.coef_):
    print(f"      + ({c:.6f})·{f}")

# ============================================================
# 3. МОДЕЛЬ 2: Линейная без свободного члена
# ============================================================
print("\n" + "─"*70)
print("МОДЕЛЬ 2: Линейная без свободного члена")
print("─"*70)

m2 = LinearRegression(fit_intercept=False).fit(X, y)
y2 = m2.predict(X)
print(f"R²  = {r2_score(y, y2):.6f}")
print(f"MAE = {mean_absolute_error(y, y2):.4f}")
print(f"RMSE= {np.sqrt(mean_squared_error(y, y2)):.4f}")
print(f"\nI_p = ({m2.coef_[0]:.6f})·D_n + ({m2.coef_[1]:.8f})·p_nom "
      f"+ ({m2.coef_[2]:.6f})·bar_e_0")

# ============================================================
# 4. МОДЕЛЬ 3: Только D_n (главный фактор)
# ============================================================
print("\n" + "─"*70)
print("МОДЕЛЬ 3: Только через D_n (без свободного члена)")
print("─"*70)

X_dn = df[['D_n']].values
m3 = LinearRegression(fit_intercept=False).fit(X_dn, y)
y3 = m3.predict(X_dn)
print(f"R²  = {r2_score(y, y3):.6f}")
print(f"MAE = {mean_absolute_error(y, y3):.4f}")
print(f"\nI_p = ({m3.coef_[0]:.6f})·D_n")

# ============================================================
# 5. МОДЕЛЬ 4: Степенная (степенной закон) — самое физичное
#    I_p = C · D_n^a · p_nom^b · bar_e_0^c
#    Логарифмируем и строим линейную регрессию
# ============================================================
print("\n" + "─"*70)
print("МОДЕЛЬ 4: Степенная  I_p = C · D_n^a · p_nom^b · bar_e_0^c")
print("─"*70)

log_X = np.log(df[features].values)
log_y = np.log(y)

m4 = LinearRegression().fit(log_X, log_y)
log_y4 = m4.predict(log_X)
y4 = np.exp(log_y4)

# R² считаем в исходных координатах
r2_4 = r2_score(y, y4)
print(f"R²  (в исходных) = {r2_4:.6f}")
print(f"R²  (в логарифмах) = {r2_score(log_y, log_y4):.6f}")
print(f"MAE = {mean_absolute_error(y, y4):.4f}")
print(f"RMSE= {np.sqrt(mean_squared_error(y, y4)):.4f}")
print(f"\nln(I_p) = {m4.intercept_:.6f}")
for f, c in zip(features, m4.coef_):
    print(f"         + ({c:.6f})·ln({f})")
print(f"\nC = exp({m4.intercept_:.6f}) = {np.exp(m4.intercept_):.6f}")
print(f"\nI_p = {np.exp(m4.intercept_):.6f} · "
      f"D_n^{m4.coef_[0]:.6f} · "
      f"p_nom^{m4.coef_[1]:.8f} · "
      f"bar_e_0^{m4.coef_[2]:.6f}")

# ============================================================
# 6. МОДЕЛЬ 5: Полином 2-й степени (с взаимодействиями)
# ============================================================
print("\n" + "─"*70)
print("МОДЕЛЬ 5: Полином 2-й степени с взаимодействиями")
print("─"*70)

poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)
names = poly.get_feature_names_out(features)

m5 = LinearRegression().fit(X_poly, y)
y5 = m5.predict(X_poly)
print(f"R²  = {r2_score(y, y5):.6f}")
print(f"MAE = {mean_absolute_error(y, y5):.4f}")
print(f"RMSE= {np.sqrt(mean_squared_error(y, y5)):.4f}")
print(f"\nI_p = {m5.intercept_:.6f}")
for n, c in zip(names, m5.coef_):
    sign = '+' if c >= 0 else '-'
    print(f"      {sign} ({abs(c):.6f})·{n}")

# ============================================================
# 7. СВОДНАЯ ТАБЛИЦА
# ============================================================
print("\n" + "="*70)
print("СВОДНАЯ ТАБЛИЦА КАЧЕСТВА МОДЕЛЕЙ")
print("="*70)

results = pd.DataFrame({
    'Модель': [
        'Линейная',
        'Линейная без своб. члена',
        'Только D_n',
        'Степенная',
        'Полином 2-й степени'
    ],
    'R²': [
        r2_score(y, y1),
        r2_score(y, y2),
        r2_score(y, y3),
        r2_4,
        r2_score(y, y5)
    ],
    'MAE': [
        mean_absolute_error(y, y1),
        mean_absolute_error(y, y2),
        mean_absolute_error(y, y3),
        mean_absolute_error(y, y4),
        mean_absolute_error(y, y5)
    ],
    'RMSE': [
        np.sqrt(mean_squared_error(y, y1)),
        np.sqrt(mean_squared_error(y, y2)),
        np.sqrt(mean_squared_error(y, y3)),
        np.sqrt(mean_squared_error(y, y4)),
        np.sqrt(mean_squared_error(y, y5))
    ]
})
print(results.round(6).to_string(index=False))

# ============================================================
# 8. ПРОВЕРКА НА НЕСКОЛЬКИХ ТОЧКАХ
# ============================================================
print("\n" + "="*70)
print("ПРОВЕРКА: сравнение факт vs модель (степенная)")
print("="*70)

check = df[['Name', 'D_n', 'p_nom', 'bar_e_0', 'I_p']].copy()
check['I_p_модель'] = y4
check['откл., %'] = 100 * (check['I_p_модель'] - check['I_p']) / check['I_p']
print(check.head(20).to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# 1. Загрузка данных
# ============================================================
df = pd.read_excel('database-1-2026-09-24.xlsx', sheet_name='Лист1')

target = 'I_p'
x_col = 'D_n'
group_col = 'p_nom'
fixed_col = 'bar_e_0'

p_nom_values = sorted(df[group_col].unique())
bar_e_values = sorted(df[fixed_col].unique())

# ============================================================
# 2. МНК-модель (степенная):  I_p = C·D_n^a·p_nom^b·bar_e_0^c
# ============================================================
X = df[['D_n', 'p_nom', 'bar_e_0']].values
y = df[target].values

log_X = np.log(X)
log_y = np.log(y)

m_mnk = LinearRegression().fit(log_X, log_y)
C = np.exp(m_mnk.intercept_)
a, b, c = m_mnk.coef_

y_mnk = np.exp(m_mnk.predict(log_X))
r2_mnk = r2_score(y, y_mnk)

print("="*70)
print("МНК-МОДЕЛЬ (степенная)")
print("="*70)
print(f"I_p = {C:.6f} · D_n^{a:.6f} · p_nom^{b:.8f} · bar_e_0^{c:.6f}")
print(f"R² = {r2_mnk:.6f}")

def I_p_mnk(D_n, p_nom, bar_e_0):
    return C * (D_n ** a) * (p_nom ** b) * (bar_e_0 ** c)

# ============================================================
# 3. ФУНКЦИОНАЛЬНЫЕ кривые — полином 2-й степени,
#    построенный ОТДЕЛЬНО для каждой серии (p_nom, bar_e_0)
# ============================================================
func_models = {}   # ключ (p_nom, bar_e_0) -> (poly, model, r2)

for p_val in p_nom_values:
    for be_val in bar_e_values:
        sub = df[(df[group_col] == p_val) &
                 (df[fixed_col] == be_val)].sort_values(x_col)
        if len(sub) < 3:
            continue
        xg = sub[x_col].values.reshape(-1, 1)
        yg = sub[target].values

        poly = PolynomialFeatures(degree=2, include_bias=False)
        xg_poly = poly.fit_transform(xg)
        m = LinearRegression().fit(xg_poly, yg)
        r2 = r2_score(yg, m.predict(xg_poly))
        func_models[(p_val, be_val)] = (poly, m, r2)

print("\nФункциональные кривые (полином 2-й степени) — R² по сериям:")
for k, v in func_models.items():
    print(f"  p_nom={k[0]:.0f}, bar_e_0={k[1]}: R² = {v[2]:.6f}")

def I_p_func(D_n, p_nom, bar_e_0):
    """Возвращает значения функциональной кривой для массива D_n."""
    key = (p_nom, bar_e_0)
    if key not in func_models:
        return None
    poly, m, _ = func_models[key]
    D = np.asarray(D_n).reshape(-1, 1)
    return m.predict(poly.transform(D))

# ============================================================
# 4. Настройка фигуры
# ============================================================
fig, ax = plt.subplots(figsize=(11, 7))

x_min, x_max = df[x_col].min(), df[x_col].max()
y_min, y_max = df[target].min(), df[target].max()
x_pad = 0.05 * (x_max - x_min)
y_pad = 0.08 * (y_max - y_min)

D_grid = np.linspace(x_min, x_max, 400)

colors = plt.cm.plasma(np.linspace(0.1, 0.85, len(p_nom_values)))
color_map = {p: colors[i] for i, p in enumerate(p_nom_values)}

# ============================================================
# 5. Функция анимации
# ============================================================
def update(frame):
    ax.clear()
    be_val = bar_e_values[frame]

    for p_val in p_nom_values:
        col = color_map[p_val]

        # --- 1) МНК-кривая (степенная, сплошная) ---
        y_mnk_curve = I_p_mnk(D_grid, p_val, be_val)
        ax.plot(D_grid, y_mnk_curve, '-',
                color=col, linewidth=2.4,
                label=f'$p_{{ном}}$ = {p_val/1e6:.0f} МПа — МНК')

        # --- 2) Функциональная кривая (полином, штриховая) ---
        y_func_curve = I_p_func(D_grid, p_val, be_val)
        if y_func_curve is not None:
            ax.plot(D_grid, y_func_curve, '--',
                    color=col, linewidth=1.8, alpha=0.9,
                    label=f'$p_{{ном}}$ = {p_val/1e6:.0f} МПа — функц.')

        # --- 3) Исходные точки ---
        sub = df[(df[group_col] == p_val) &
                 (df[fixed_col] == be_val)].sort_values(x_col)
        if len(sub):
            ax.scatter(sub[x_col], sub[target],
                       s=90, facecolors='none',
                       edgecolors=col, linewidths=2.2,
                       marker='o', zorder=5,
                       label=f'$p_{{ном}}$ = {p_val/1e6:.0f} МПа— данные')

    ax.set_xlim(x_min - x_pad, x_max + x_pad)
    ax.set_ylim(y_min - y_pad, y_max + y_pad)
    ax.set_xlabel('$D_n$, м', fontsize=16)
    ax.set_ylabel('$I_p$, кН$\cdot$с', fontsize=16)
    ax.set_title(
        f'$I_p = {C:.4f} \\cdot D_н^{{{a:.4f}}} \\cdot '
        f'p_{{ном}}^{{{b:.6f}}} \\cdot \\bar{{e}}_0^{{{c:.4f}}}$ \n'
        f'$\\bar{{e}}_0$ = {be_val}',
        fontsize=13)
    ax.grid(alpha=0.3)
    ax.legend(fontsize=8.5, loc='upper left', ncol=1)

# ============================================================
# 6. Создание и сохранение GIF
# ============================================================
ani = animation.FuncAnimation(
    fig, update,
    frames=len(bar_e_values),
    interval=1600,
    repeat=True
)

ani.save('Result_I_p_var_p_nom.gif', writer='pillow', fps=1, dpi=120)
print("\nСохранено: Result_I_p_var_p_nom.gif")

plt.close()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# 1. Загрузка данных
# ============================================================
df = pd.read_excel('database-1-2026-09-24.xlsx', sheet_name='Лист1')

target = 'I_p'
x_col = 'D_n'
group_col = 'bar_e_0'    # кривые на кадре — разные bar_e_0
fixed_col = 'p_nom'      # кадры — фиксируем p_nom

p_nom_values = sorted(df[fixed_col].unique())
bar_e_values = sorted(df[group_col].unique())

# ============================================================
# 2. МНК-модель (степенная):  I_p = C·D_n^a·p_nom^b·bar_e_0^c
# ============================================================
X = df[['D_n', 'p_nom', 'bar_e_0']].values
y = df[target].values

log_X = np.log(X)
log_y = np.log(y)

m_mnk = LinearRegression().fit(log_X, log_y)
C = np.exp(m_mnk.intercept_)
a, b, c = m_mnk.coef_

r2_mnk = r2_score(y, np.exp(m_mnk.predict(log_X)))

print("="*70)
print("МНК-МОДЕЛЬ (степенная, единая для всех данных)")
print("="*70)
print(f"I_p = {C:.6f} · D_n^{a:.6f} · p_nom^{b:.8f} · bar_e_0^{c:.6f}")
print(f"R² = {r2_mnk:.6f}")

def I_p_mnk(D_n, p_nom, bar_e_0):
    return C * (D_n ** a) * (p_nom ** b) * (bar_e_0 ** c)

# ============================================================
# 3. ФУНКЦИОНАЛЬНЫЕ кривые — полином 2-й степени,
#    ОТДЕЛЬНО для каждой серии (p_nom, bar_e_0)
# ============================================================
func_models = {}

for p_val in p_nom_values:
    for be_val in bar_e_values:
        sub = df[(df[group_col] == be_val) &
                 (df[fixed_col] == p_val)].sort_values(x_col)
        if len(sub) < 3:
            continue
        xg = sub[x_col].values.reshape(-1, 1)
        yg = sub[target].values

        poly = PolynomialFeatures(degree=2, include_bias=False)
        xg_poly = poly.fit_transform(xg)
        m = LinearRegression().fit(xg_poly, yg)
        r2 = r2_score(yg, m.predict(xg_poly))
        func_models[(p_val, be_val)] = (poly, m, r2)

print("\nФункциональные кривые (полином 2-й степени):")
for k, v in func_models.items():
    print(f"  p_nom={k[0]:.0f}, bar_e_0={k[1]}: R² = {v[2]:.6f}")

def I_p_func(D_n, p_nom, bar_e_0):
    key = (p_nom, bar_e_0)
    if key not in func_models:
        return None
    poly, m, _ = func_models[key]
    D = np.asarray(D_n).reshape(-1, 1)
    return m.predict(poly.transform(D))

# ============================================================
# 4. Настройка фигуры
# ============================================================
fig, ax = plt.subplots(figsize=(11, 7))

x_min, x_max = df[x_col].min(), df[x_col].max()
y_min, y_max = df[target].min(), df[target].max()
x_pad = 0.05 * (x_max - x_min)
y_pad = 0.08 * (y_max - y_min)

D_grid = np.linspace(x_min, x_max, 400)

colors = plt.cm.viridis(np.linspace(0.15, 0.85, len(bar_e_values)))
color_map = {be: colors[i] for i, be in enumerate(bar_e_values)}

# ============================================================
# 5. Функция анимации
# ============================================================
def update(frame):
    ax.clear()
    p_val = p_nom_values[frame]           # фиксируем p_nom
    sub = df[df[fixed_col] == p_val]      # фильтр по p_nom

    for be_val in bar_e_values:
        col = color_map[be_val]

        # --- 1) МНК-кривая (степенная, сплошная) ---
        y_mnk_curve = I_p_mnk(D_grid, p_val, be_val)
        ax.plot(D_grid, y_mnk_curve, '-',
                color=col, linewidth=2.4,
                label=f'$\\bar{{e}}_0$ = {be_val} — МНК')

        # --- 2) Функциональная кривая (полином, штриховая) ---
        y_func_curve = I_p_func(D_grid, p_val, be_val)
        if y_func_curve is not None:
            ax.plot(D_grid, y_func_curve, '--',
                    color=col, linewidth=1.8, alpha=0.9,
                    label=f'$\\bar{{e}}_0$ = {be_val} — функц.')

        # --- 3) Исходные точки ---
        g = sub[sub[group_col] == be_val].sort_values(x_col)
        if len(g):
            ax.scatter(g[x_col], g[target],
                       s=90, facecolors='none',
                       edgecolors=col, linewidths=2.2,
                       marker='o', zorder=5,
                       label=f'$\\bar{{e}}_0$ = {be_val} — данные')

    ax.set_xlim(x_min - x_pad, x_max + x_pad)
    ax.set_ylim(y_min - y_pad, y_max + y_pad)
    ax.set_xlabel('$D_н$, м', fontsize=12)
    ax.set_ylabel('$I_p$, кН$\cdot$с', fontsize=12)
    ax.set_title(
        f'$I_p = {C:.4f} \\cdot D_н^{{{a:.4f}}} \\cdot '
    f'p_{{ном}}^{{{b:.6f}}} \\cdot \\bar{{e}}_0^{{{c:.4f}}}$ \n'
        f'$p_{{ном}}$ = {p_val/1e6:,.0f} МПа',
        fontsize=13)
    ax.grid(alpha=0.3)
    ax.legend(fontsize=8, loc='upper left', ncol=1)

# ============================================================
# 6. Создание и сохранение GIF
# ============================================================
ani = animation.FuncAnimation(
    fig, update,
    frames=len(p_nom_values),
    interval=1600,
    repeat=True
)

ani.save('Result_I_p_var_bar_e_0.gif', writer='pillow', fps=1, dpi=120)
print("\nСохранено: Result_I_p_var_bar_e_0.gif")

plt.close()
```
